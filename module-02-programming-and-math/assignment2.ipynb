**Task 1: Import and Load the Data**

What to do: Upload the Superstore CSV to Colab and load it into a Pandas DataFrame.

```python
import pandas as pd

df = pd.read_csv("Sample - Superstore.csv", encoding='latin1')
df.head()
```

**Task 2: Inspect the Structure**

What to do: Examine the dataset's shape, data types, missing values, and summary statistics.

```python
print("Shape: ", df.shape) #rows and columns
df.info() #Data types and missing values
df.describe() #Statistical summary
```

The dataset contains 9994 rows and 21 columns with no missing values across all columns. The data includes 15 object(text) columns, 3 float64 columns - notably, Order Date and Ship Date are stored as text(object) and will need to be converted to datetime format. Sales ranges from $0.44 to $22638, and Profit ranges from -$6,599 to $8,399, indicating high variability and the presence of loss-making orders.

**Task 3: Handle Missing Values**

What to do: Fill missing values in numerical columns with the median and categorical columns with the mode.


```python
from numpy import number
# Step 1: Check which columns have missing values
print("Missing values before cleaning:")
print(df.isnull().sum())

# Step 2: Fill missing values (good practice even if none exist now)
# Numerical columns → fill with median
numerical_cols = df.select_dtypes(include = 'number').columns
df[numerical_cols] = df[numerical_cols].fillna(df[numerical_cols].median())

# Categorical columns → fill with mode
categorical_cols = df.select_dtypes(include = "object").columns
for col in categorical_cols:
  df[col] = df[col].fillna(df[col].mode()[0])

# Step 3: Verify no missing values remain
print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

The dataset had no missing values, so no imputation was necessary. However, the cleaning code was applied as a precaution — numerical columns would be filled with the median to handle outlier robustness, and categorical columns with the most frequent value (mode). The final check confirms zero null values across all 21 columns.

**Task 4: Remove Duplicates**

What to do: Check for and remove duplicate rows.


```python
# Step 1: Count duplicate rows
print("Number of duplicate rows:", df.duplicated().sum)

# Step 2: Remove duplicates
df = df.drop_duplicates()

# Step 3: Report final shape
print("Rows removed:", 9994 - df.shape[0])
print("Final shape:", df.shape)
```

The dataset was checked for identical duplicate rows. After removing duplicates, the final shape is reported — if the row count matches 9,994, no duplicates existed. Data integrity is now ensured with all remaining rows being unique.

**Task 5: Convert Data Types**

What to do: Convert Order Date and Ship Date columns to the proper datetime format.


```python
# Step 1: Convert Order Date and Ship Date to datetime
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

# Step 2: Confirm the conversion
print(df[['Order Date', 'Ship Date']].dtypes)

# Step 3: Spot-check a few rows
df[['Order Date', 'Ship Date']].head(5)
```

The Order Date and Ship Date columns were originally stored as text (object type) and have been successfully converted to datetime64 format. This allows for proper date-based operations such as sorting, filtering, and calculating time differences. A spot-check of the first 5 rows confirms the dates are formatted correctly.

**Task 6: Create Derived Features**

What to do: Create three customer-level summary metrics: total spending, order frequency, and average order value.

```python
total_spending = df.groupby('Customer ID')['Sales'].sum().rename('Total Spending')

order_frequency = df.groupby('Customer ID')['Order ID'].nunique().rename('Order Frequency')

avg_order_value = df.groupby('Customer ID')['Sales'].mean().rename('Average Order Value')



customer_summary = pd.concat([total_spending, order_frequency, avg_order_value], axis=1,).reset_index()
customer_summary.head()
```

```python
# Save cleaned main dataset
df.to_csv('superstore_cleaned.csv', index = False)

# Save customer summary table
customer_summary.to_csv('customer_summary.csv', index=False)

print("Files saved successfully!")
```
